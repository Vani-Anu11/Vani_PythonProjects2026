```python
# print Odd Numbers in the list 
numbers = [20, 10, 16, 19, 25, 1, 276, 188]

for num in numbers:
    if num % 2 != 0:   # odd check
        print(num, "is odd")
```

```python
#even numbers list
numbers = [20, 10, 16, 19, 25, 1, 276, 188]

for num in numbers:
    if num % 2 == 0:   # even check
        print(num, "is even")
```

```python
#user data validation and return result
age = int(input("Enter your age: "))

if age < 13:
    print("Child")
elif 13 <= age < 20:
    print("Teenager")
elif 20 <= age < 60:
    print("Adult")
else:
    print("Senior Citizen")
```
