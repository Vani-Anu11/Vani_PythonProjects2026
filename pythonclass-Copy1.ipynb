```python
class SubfieldsInAI:
    def Subfields(self):
        subfields = [
            "Machine Learning",
            "Neural Networks",
            "Vision",
            "Robotics",
            "Speech Processing",
            "Natural Language Processing"
        ]
        print("Sub-fields in AI are:")
        for field in subfields:
            print(field)

# Run
SubfieldsInAI().Subfields()
```

```python
# In [17]: Odd or Even with input
class OddEven:
    def OddEven(self):
        num = int(input("Enter a number: "))
        if num % 2 == 0:
            print(f"{num} is Even number")
        else:
            print(f"{num} is Odd number")

# Run
OddEven().OddEven()
```

```python
# In [10]: Eligibility for Marriage
class EligibilityForMarriage:
    def Eligible(self):
        gender = input("Your Gender : ")
        age = int(input("Your Age : "))
        
        if gender.lower() == "male":
            if age >= 21:
                print("ELIGIBLE")
            else:
                print("NOT ELIGIBLE")
        elif gender.lower() == "female":
            if age >= 18:
                print("ELIGIBLE")
            else:
                print("NOT ELIGIBLE")
        else:
            print("Invalid gender")

# Run
EligibilityForMarriage().Eligible()
```

```python
# In [12]: Percentage of Marks
class FindPercent:
    def percentage(self):
        # Taking input for 5 subjects
        s1 = int(input("Subject1 = "))
        s2 = int(input("Subject2 = "))
        s3 = int(input("Subject3 = "))
        s4 = int(input("Subject4 = "))
        s5 = int(input("Subject5 = "))
        
        total = s1 + s2 + s3 + s4 + s5
        percent = (total / 500) * 100   # since each subject is out of 100
        
        print("Total :", total)
        print("Percentage :", percent)

# Run
FindPercent().percentage()
```

```python
# In [14]: Area and Perimeter of Triangle
class Triangle:
    def triangle(self):
        # Inputs for area
        height = int(input("Height : "))
        breadth = int(input("Breadth : "))
        area = (height * breadth) / 2
        print("Area formula: (Height * Breadth) / 2")
        print("Area of Triangle:", area)

        # Inputs for perimeter
        height1 = int(input("Height1 : "))
        height2 = int(input("Height2 : "))
        breadth2 = int(input("Breadth : "))
        perimeter = height1 + height2 + breadth2
        print("Perimeter formula: Height1 + Height2 + Breadth")
        print("Perimeter of Triangle:", perimeter)

# Run
Triangle().triangle()
```
